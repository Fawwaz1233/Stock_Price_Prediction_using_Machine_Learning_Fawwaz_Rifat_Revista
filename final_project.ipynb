```python
import pandas as pd
import numpy as np
```

```python
df2 = pd.read_csv('financial_data_2_income statement.csv')
```

```python
df2.head(20)
```

```python
col_index = df2.columns.get_loc('Q115') + 1
df2 = df2.iloc[:, :col_index]
```

```python
df2.drop('Unnamed: 0', axis=1, inplace=True)
```

```python
# 🔹 Daftar fitur (9 fitur per kuartal)
features = [
    "share_outstanding", "eps", "per", "price_to_sales", "ebitda",
    "roa", "roe", "roce", "interest_coverage"
]

# 🔹 Daftar kode emiten
emiten_code = [
    "BINA", "BJBR", "BJTM", "BKSW", "BMAS", "BMRI", "BNBA", "BNGA",
    "BNII", "BNLI", "BSIM", "BSWD", "BTPN", "BVIC", "DNAR", "INPC",
    "MAYA", "AGRO", "AGRS", "BABP", "BACA", "BBCA", "BBKP", "BBMD",
    "BBNI", "BBRI", "BBTN", "BBYB", "BCIC", "BDMN", "BEKS", "MCOR",
    "MEGA", "NISP", "NOBU", "PNBN", "PNBS", "SDRA"
]

result = []
num_emitens = len(emiten_code)  # Jumlah kode emiten
num_rows_per_emiten = len(df2) // (9 * num_emitens)  # Banyaknya data per emiten

for col in df2.columns:
    for i in range(num_emitens):
        for start in range(i * 9 * num_rows_per_emiten, (i + 1) * 9 * num_rows_per_emiten, 9):
            if start + 9 <= len(df2):
                row = {feature: df2[col].iloc[start + j] for j, feature in enumerate(features)}
                row['date'] = col
                row['emiten_code'] = emiten_code[i]  # Tambahkan kode emiten
                result.append(row)

# 🔹 Mengubah hasil menjadi DataFrame baru
final_df2 = pd.DataFrame(result)

# 🔹 Tampilkan hasil akhir
print(final_df2)
```

```python
final_df2
```

```python
final_df2 = final_df2.drop(columns=['share_outstanding','price_to_sales','roce','interest_coverage'])
```
